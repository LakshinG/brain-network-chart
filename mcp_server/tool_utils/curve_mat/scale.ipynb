```python
import h5py
import numpy as np

input_path = "/ram/USERS/tao/code/gift/BrainChart-FC-Lifespan/brain_network_app/Lifespan/curve_mat/CT.mat"
output_path = "/ram/USERS/tao/code/gift/BrainChart-FC-Lifespan/brain_network_app/Lifespan/curve_mat/CT_0_20.mat"

with h5py.File(input_path, "r") as f:
    centiles = f["centiles"][:]      # (5, 200)
    age = f["X"][:]                # (1, 200)

# 把 age 变成一维
age = age.flatten()                  # (200,)

# 生成 mask
mask = (age >= 0) & (age <= 20)

# 筛选
centiles_filtered = centiles[:, mask]
age_filtered = age[mask]

# 保存
with h5py.File(output_path, "w") as f:
    f.create_dataset("centiles", data=centiles_filtered)
    f.create_dataset("X", data=age_filtered.reshape(1, -1))

print("Saved 0–20 years subset.")
```
